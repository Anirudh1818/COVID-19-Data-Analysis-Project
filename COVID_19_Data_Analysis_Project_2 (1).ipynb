## 12. Find the five countries with the highest number of COVID-19 cases

```python
highest_cases = df.nlargest(5, "Total Cases")[["Country", "Total Cases"]]
highest_cases
```

**Observation:** The USA has the highest total number of reported cases in this dataset, followed by India, Brazil, France and Germany. These five countries make up the top group when comparing total cases.

## 13. Create a bar plot of the five countries with the highest total cases

```python
plt.figure(figsize=(10, 5))
plt.bar(highest_cases["Country"], highest_cases["Total Cases"])
plt.title("Top 5 Countries with Highest COVID-19 Cases")
plt.xlabel("Country")
plt.ylabel("Total Cases")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

**Observation:** The bar chart makes the difference between the five countries easier to see. The USA has a noticeably higher case count than the other countries shown.

## 14. Find the five countries with the lowest number of COVID-19 cases

## 15. Create a bar plot of the five countries with the lowest total cases

**Observation:** The graph shows how low the case counts are for these five entries. The values are very small in comparison with the top five countries.

## 19. Find the five countries with the highest COVID-19 death percentage

## 20. Create a bar plot of the five countries with the highest death percentage

## 21. Calculate total COVID-19 cases continent-wise using groupby()

```python
continent_cases = df.groupby("Continent")["Total Cases"].sum()
continent_cases
```

**Observation:** Europe has the highest combined total cases in this dataset, followed by Asia and Northern America. Oceania has the lowest combined total among the listed continents.

## 22. Reset the index and create a bar plot of total cases by continent

```python
continent_cases_df = continent_cases.reset_index()
plt.figure(figsize=(11, 5))
plt.bar(continent_cases_df["Continent"], continent_cases_df["Total Cases"])
plt.title("Total COVID-19 Cases by Continent")
plt.xlabel("Continent")
plt.ylabel("Total Cases")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

**Observation:** The continent-wise chart gives a clearer comparison of the combined case totals. Europe and Asia have much higher totals than Africa and Oceania in this dataset.

## 23. Create a scatter plot showing the relationship between Population and Total Cases

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Population", y="Total Cases")
plt.title("Population vs Total COVID-19 Cases")
plt.xlabel("Population")
plt.ylabel("Total Cases")
plt.tight_layout()
plt.show()
```

**Observation:** The scatter plot shows that countries with larger populations often have higher total case counts, although this is not true for every country. Population alone does not explain the number of cases.

## 24. Create a scatter plot showing the relationship between Total Cases and Total Deaths

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Total Cases", y="Total Deaths")
plt.title("Total COVID-19 Cases vs Total Deaths")
plt.xlabel("Total Cases")
plt.ylabel("Total Deaths")
plt.tight_layout()
plt.show()
```

**Observation:** The points generally move upward as total cases increase, which suggests a positive relationship between total cases and total deaths in this dataset. There are still differences between countries.

## 25. Calculate Spearman rank correlation and visualize it using a heatmap

```python
numeric_df = df.select_dtypes(include="number")
spearman_corr = numeric_df.corr(method="spearman")
spearman_corr
```

**Observation:** The Spearman correlation values show how strongly numerical columns move together in terms of their ranks. Values closer to 1 indicate a positive relationship, values closer to -1 indicate a negative relationship, and values near 0 indicate a weak relationship.

```python
plt.figure(figsize=(10, 7))
sns.heatmap(spearman_corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Spearman Rank Correlation Heatmap")
plt.tight_layout()
plt.show()
```

**Observation:** The heatmap makes it easier to compare relationships across all numerical columns. In this dataset, total cases and total deaths have a strong positive relationship. Population and total deaths also show a positive relationship, while some per-million measures have weaker or negative relationships with population.
